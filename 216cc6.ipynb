```python
import nltk
import random
import string
import numpy as np

nltk.download('punkt')
nltk.download('wordnet')

from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

lemmatizer = WordNetLemmatizer()

corpus = {
    "greetings": ["hello", "hi", "howdy", "hey"],
    "goodbyes": ["bye", "goodbye", "see you later", "take care"],
    "thanks": ["thank you", "thanks", "appreciate it"],
    "how_are_you": ["how are you?", "how's it going?", "what's up?"],
    "default": ["I'm sorry, I don't understand."]
}

# The 'default' key was missing in the original responses dictionary.
# Adding it here to handle cases where no matching phrase is found in the corpus.
responses = {
    "greetings": ["Hello! How can I assist you?", "Hi there! What can I do for you?", "Hey! How's it going?"],
    "goodbyes": ["Goodbye! Have a great day!", "See you later!", "Take care!"],
    "thanks": ["You're welcome!", "No problem!", "Glad to help!"],
    "how_are_you": ["I'm just a bot, but I'm here to help you!", "I'm doing great, thanks for asking!"],
    "default": ["I'm sorry, I don't understand."]
}

def preprocess_input(user_input):
    tokens = word_tokenize(user_input.lower())
    return [lemmatizer.lemmatize(token) for token in tokens]

def generate_response(user_input):
    processed_input = preprocess_input(user_input)

    for key, phrases in corpus.items():
        if any(phrase in user_input for phrase in phrases):
            return random.choice(responses[key])

    return random.choice(responses["default"])

def chatbot():
    print("Chatbot: Hi! I'm your assistant. Type 'exit' to end the chat.")
    while True:
        user_input = input("You: ")
        if user_input.lower() == 'exit':
            print("Chatbot: Goodbye!")
            break

        response = generate_response(user_input)
        print("Chatbot:", response)

if __name__ == "__main__":
    chatbot()
```
